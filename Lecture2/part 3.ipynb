```python
from countries import countriesWWWWW
from countries_details_dat import countries_data


land_countries = [country for country in countriesWWWWW if 'land' in country]
print(land_countries)
print()


all_languages = set()
for country in countries_data:
    all_languages.update(country['languages'])

print(len(all_languages))
print()


language_counts = {}
for country in countries_data:
    for lang in country['languages']:
        language_counts[lang] = language_counts.get(lang, 0) + 1

top_10_languages = sorted(language_counts, key=language_counts.get, reverse=True)[:10]
print(top_10_languages)
print()


top_10_population = sorted(countries_data, key=lambda c: c['population'], reverse=True)[:10]
for c in top_10_population:
    print(c['name'], c['population'])
```
