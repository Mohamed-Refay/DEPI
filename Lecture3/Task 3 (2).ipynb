```python
def add_all_nums(*args):
    for item in args:
        if not isinstance(item, (int, float)):
            return f"Error: '{item}' is not a number"
    return sum(args)
```

## Q6

```python
import math

def solve_quadratic_eqn(a, b, c):
    d = b**2 - 4*a*c
    if d < 0:
        return "No real solutions"
    elif d == 0:
        x = -b / (2*a)
        return [x]
    else:
        x1 = (-b + math.sqrt(d)) / (2*a)
        x2 = (-b - math.sqrt(d)) / (2*a)
        return [x1, x2]
```

```python
print(solve_quadratic_eqn(1, -3, 2))
print(solve_quadratic_eqn(1, 2, 1))
print(solve_quadratic_eqn(1, 0, 1))
```

```python
def reverse_list(arr):
    reversed_arr = []
    for i in range(len(arr) - 1, -1, -1):
        reversed_arr.append(arr[i])
    return reversed_arr

print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C"]))
```

## Q4

```python
def evens_and_odds(n):
    evens = 0
    odds = 0
    for i in range(0, n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1
    return f"The number of odds are {odds}.\nThe number of evens are {evens}."

print(evens_and_odds(100))
```

## Q5

```python
def calculate_mean(lst):
    return sum(lst) / len(lst)

def calculate_median(lst):
    sorted_lst = sorted(lst)
    n = len(sorted_lst)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_lst[mid - 1] + sorted_lst[mid]) / 2
    else:
        return sorted_lst[mid]

def calculate_mode(lst):
    from collections import Counter
    counts = Counter(lst)
    max_count = max(counts.values())
    modes = [k for k, v in counts.items() if v == max_count]
    if len(modes) == 1:
        return modes[0]
    return modes

data = [2, 4, 4, 4, 5, 5, 7, 9]
print(calculate_mean(data))
print(calculate_median(data))
print(calculate_mode(data))
```

## Q1

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(10))
print(is_prime(2))
```

## Q2

```python
def all_unique(lst):
    return len(lst) == len(set(lst))

print(all_unique([1, 2, 3, 4]))
print(all_unique([1, 2, 2, 4]))
```

## Q3

```python
def all_same_type(lst):
    if len(lst) == 0:
        return True
    first_type = type(lst[0])
    return all(type(item) == first_type for item in lst)

print(all_same_type([1, 2, 3]))
print(all_same_type([1, "2", 3]))
```

## Q4

```python
import keyword

def is_valid_variable(name):
    return isinstance(name, str) and name.isidentifier() and not keyword.iskeyword(name)

print(is_valid_variable("my_var"))
print(is_valid_variable("2var"))
print(is_valid_variable("my var"))
print(is_valid_variable("class"))
```
