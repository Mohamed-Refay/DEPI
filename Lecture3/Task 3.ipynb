## Q2

```python
def add_all_nums(*args):
    for item in args:
        if not isinstance(item, (int, float)):
            return f"Error: '{item}' is not a number"
    return sum(args)
```

```python
print(add_all_nums(1, 2, 3))
print(add_all_nums(1, "x", 3))
```

## Q4

```python
def check_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"
```

```python
print(check_season(1))
print(check_season(4))
print(check_season(7))
print(check_season(10))
```

## Q6

```python
import math

def solve_quadratic_eqn(a, b, c):
    d = b**2 - 4*a*c
    if d < 0:
        return "No real solutions"
    elif d == 0:
        x = -b / (2*a)
        return [x]
    else:
        x1 = (-b + math.sqrt(d)) / (2*a)
        x2 = (-b - math.sqrt(d)) / (2*a)
        return [x1, x2]
```

```python
print(solve_quadratic_eqn(1, -3, 2))
print(solve_quadratic_eqn(1, 2, 1))
print(solve_quadratic_eqn(1, 0, 1))
```

## Q10

```python
def remove_item(lst, item):
    new_list = lst.copy()
    new_list.remove(item)
    return new_list
```

```python
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))

numbers = [2, 3, 7, 9]
print(remove_item(numbers, 3))
```

## Q12

```python
def sum_of_odds(n):
    return sum(x for x in range(1, n + 1) if x % 2 != 0)
```

```python
print(sum_of_odds(5))
print(sum_of_odds(10))
```
