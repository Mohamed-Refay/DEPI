### Task 1 — Load and Explore the CSV

**1. Load `nlp_customer_messages.csv` into a DataFrame called `df`**

```python
import pandas as pd

pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", 250)

df = pd.read_csv("nlp_customer_messages.csv")
```

**2. Print the first 5 messages, shape, and column names**

```python
print(df.head(5))
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

**3. Check missing values in each column**

```python
print(df.isnull().sum())
```

**4. Check duplicated `message_id` values**

```python
print("Duplicated message_id:", df["message_id"].duplicated().sum())
```

**5. Find the longest and shortest messages based on character count**

```python
lengths = df["message"].str.len()
longest = df.loc[lengths.idxmax()]
shortest = df.loc[lengths.idxmin()]

print("Longest :", longest["message_id"], "-", lengths.max(), "characters")
print(longest["message"])
print()
print("Shortest:", shortest["message_id"], "-", lengths.min(), "characters")
print(shortest["message"])
```

**6. Create `char_count` and explain what it measures**

```python
df["char_count"] = df["message"].str.len()
print(df[["message_id", "message", "char_count"]].head(5))
```

`char_count` is the number of characters in each message. It counts letters, digits, spaces, punctuation, and emojis. It is not the number of words.
